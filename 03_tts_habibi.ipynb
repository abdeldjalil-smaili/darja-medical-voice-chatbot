```python
# Cellule 3 — Générer les 5 phrases
import os

REF_AUDIO = "ref_voice.wav"
REF_TEXT  = "نيفي راهو يسيل"

phrases = {
    "phrase1": "عندك الوجع في راسي ؟ خود هاد الدوا مرتين في اليوم وشرب الماء بزاف",
    "phrase2": " عندك حمى؟ لازمك ترتاح وتشرب الماء بزاف",
    "phrase3": "أعراض السكري هوما العطش الشديد والبول بزاف والتعب",
    "phrase4": "خود الدوا مع الماكلة وما تنساش تروح عند الطبيب",
    "phrase5": "الضغط المرتفع خطير، لازمك تراقب ماكلتك وتدير رياضة"
}

os.makedirs("tts_outputs", exist_ok=True)

for name, text in phrases.items():
    output_dir = f"tts_outputs/{name}"
    os.makedirs(output_dir, exist_ok=True)
    cmd = f'habibi-tts_infer-cli \
        --ref_audio "{REF_AUDIO}" \
        --ref_text "{REF_TEXT}" \
        --gen_text "{text}" \
        --dialect ALG \
        --output_dir "{output_dir}"'
    print(f"\n🎙️ Génération {name}...")
    os.system(cmd)
    print(f"✅ {name} généré")

print("\n✅ Toutes les phrases générées !")
```

```python
# Cellule 4 — Écouter les phrases
import IPython.display as ipd
import os

phrases_labels = {
    "phrase1": "عندك ألم في الرأس؟ خود هاد الدوا مرتين في اليوم وشرب الماء بزاف",
    "phrase2": "واش عندك حمى؟ لازمك ترتاح وتشرب الماء بزاف",
    "phrase3": "أعراض السكري هوما العطش الشديد والبول بزاف والتعب",
    "phrase4": "خود الدوا مع الماكلة وما تنساش تروح عند الطبيب",
    "phrase5": "الضغط المرتفع خطير، لازمك تراقب ماكلتك وتدير رياضة"
}

for name, text in phrases_labels.items():
    audio_path = f"tts_outputs/{name}/infer_cli_example.wav"
    if os.path.exists(audio_path):
        print(f"\n{'='*50}")
        print(f"📢 {name}: {text}")
        display(ipd.Audio(audio_path))
    else:
        print(f"❌ {name} non trouvé")
```
